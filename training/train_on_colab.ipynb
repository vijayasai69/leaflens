# LeafLens: field photo fine-tuning

This Colab notebook fine-tunes the existing 39-class model with labeled PlantDoc field photos plus PlantVillage images. It measures the existing model and candidate on separate held-out PlantDoc and PlantVillage sets. The script only recommends publishing when field macro-F1 improves and PlantVillage accuracy drops by no more than one percentage point.

1. Select **Runtime → Change runtime type → T4 GPU** (or another GPU).
2. Run the install cell, then upload `training/train.py` from the project.
3. Run the training cell and review the report before publishing a new checkpoint.
